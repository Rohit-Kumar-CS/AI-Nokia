```python
import pandas as pd
import numpy as np

rows = 5_000_000
np.random.seed(42)

data = {
    "customer_id": np.arange(1, rows + 1),
    "data_used_gb": np.random.uniform(0.1, 100, size=rows).round(2),
    "call_duration": np.random.randint(1, 500, size=rows),
    "region": np.random.choice(
        ["Delhi", "Mumbai", "Chennai", "Kolkata", "Bangalore"],
        size=rows
    )
}

df = pd.DataFrame(data)
df.to_csv("big_telecom_data.csv", index=False)

print("✅ big_telecom_data.csv created successfully with 5 million rows.")
```

```python
import pandas as pd

chunks = pd.read_csv("big_telecom_data.csv", chunksize=100000)

total_usage = 0
heavy_users = 0

for chunk in chunks:
    total_usage += chunk["data_used_gb"].sum()
    heavy_users += (chunk["data_used_gb"] > 50).sum()

print(f"✅ Total data usage: {total_usage:,.2f} GB")
print(f"✅ Heavy users (>50GB): {heavy_users:,}")
```

```python
import pandas as pd
from tqdm import tqdm

chunks = pd.read_csv("big_telecom_data.csv", chunksize=100000)

total_usage = 0
heavy_users = 0

for chunk in tqdm(chunks, desc="Processing..."):
    total_usage += chunk["data_used_gb"].sum()
    heavy_users += (chunk["data_used_gb"] > 50).sum()

print(f"\n✅ Total Data Usage: {total_usage:,.2f} GB")
print(f"✅ Heavy Users (>50GB): {heavy_users:,}")
```

```python
print(
    "Average data used per user (approx.):",
    round(total_usage / 5_000_000, 2),
    "GB"
)
```

```python
import pandas as pd

chunks = pd.read_csv("big_telecom_data.csv", chunksize=100000)

region_usage = {}

for chunk in chunks:
    usage = chunk.groupby("region")["data_used_gb"].sum()

    for region, total in usage.items():
        region_usage[region] = region_usage.get(region, 0) + total

print("Region-wise Total Usage:")

for region, total in region_usage.items():
    print(region, ":", round(total, 2), "GB")
```

```python

import pandas as pd

chunks = pd.read_csv("big_telecom_data.csv", chunksize=100000)

first_chunk = True

for chunk in chunks:
    heavy = chunk[chunk["data_used_gb"] > 50]

    heavy.to_csv(
        "heavy_users.csv",
        mode="a",
        index=False,
        header=first_chunk
    )

    first_chunk = False

print("✅ heavy_users.csv created successfully.")
```
