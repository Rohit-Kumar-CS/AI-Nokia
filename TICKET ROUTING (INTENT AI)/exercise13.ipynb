```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("telecom_master.csv")

print("Dataset Loaded Successfully")
print(df.head())
```

```python
total_calls = df["avg_voice_min"].sum()

plt.figure(figsize=(5,5))
plt.bar(["Total Calls"], [total_calls])
plt.title("Total Calls Made in Dataset")
plt.ylabel("Calls Made")
plt.show()
```

```python
usage_region = df.groupby("region")["avg_monthly_gb"].mean()

plt.figure(figsize=(7,5))
usage_region.plot(kind="bar")

plt.title("Average Usage (GB) by Region")
plt.xlabel("Region")
plt.ylabel("Average Usage (GB)")
plt.xticks(rotation=0)
plt.show()
```

```python
arpu_region = df.groupby("region")["arpu"].mean()

plt.figure(figsize=(7,5))
arpu_region.plot(kind="bar")

plt.title("Average ARPU by Region")
plt.xlabel("Region")
plt.ylabel("Average ARPU (INR)")
plt.xticks(rotation=0)
plt.show()
```

```python
contract_count = df["contract_type"].value_counts()

plt.figure(figsize=(6,6))
contract_count.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customers by Contract Type")
plt.ylabel("")
plt.show()
```

```python
revenue_region = df.groupby("region")["total_charges"].sum()

plt.figure(figsize=(7,5))
revenue_region.plot(kind="bar")

plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=0)
plt.show()
```

```python
churn_region = df.groupby("region")["churn"].sum()

plt.figure(figsize=(7,5))
churn_region.plot(kind="bar")

plt.title("Churn Customers by Region")
plt.xlabel("Region")
plt.ylabel("Number of Churn Customers")
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_revenue = (
    df.groupby("tenure_months")["monthly_charges"]
      .sum()
      .sort_index()
)

plt.figure(figsize=(8,5))
monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (INR)")
plt.grid(True)
plt.show()
```
